```python
import requests
from bs4 import BeautifulSoup
import json

def get_wikipedia_page(topic):
  url = f"https://en.wikipedia.org/wiki/{topic.replace(' ', '_')}"
  headers = {"User-Agent": "Mozilla/5.0"}   # Avoid blocking (403)


  response = requests.get(url, headers=headers)
  if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')
    return soup
  else:
    print(f"Failed to retrieve the page. Status code: {response.status_code}")
    return None

def get_article_title(soup):
  return soup.find('h1').text

def get_article_summary(soup):
    paragraph = soup.find_all('p')
    summary=""
    for paragraph in paragraph[:3]:
        summary += paragraph.text.strip() +" "
    return summary.strip()


def get_article_headings(soup):
    headings = soup.find_all(['h2','h3','h4','h5'])
    headings_list = []
    for heading in headings:
        headings_list.append(heading.text.replace("[edit]"," ").strip())
    return headings_list



def get_article_links(soup):
    links = soup.find_all('a',href=True)
    article_links = []
    for link in links:
      href = link['href']
      if link['href'].startswith('/wiki/') and ":" not in href:
            article_links.append(f"https://en.wikipedia.org{href}")
    return list(set(article_links))[:5]

def add_all_in_file(topic,title,summary,headings,links):


    with open(f"{topic}.txt", "w", encoding="utf-8") as file:
      file.write(f"Article Title: {title}\n\n")
      file.write(f"Article Summary: {summary}\n\n")

      file.write(f"Article Headings:\n")
      for heading in headings:
          file.write(f"- {heading}\n")

      file.write(f"\nArticle Links: \n")
      for link in links:
          file.write(f"- {link}\n")
    print("Data has been added successfully.")

def add_article_in_json(topic,title,summary,headings,links):
  data = {
        "Article Title": title,
        "Article Summary": summary,
        "Article Headings": headings,
        "Article Links": links
    }
  with open(f"{topic}.json", "w",encoding="utf-8") as file:
    json.dump(data, file, indent=4,)
print("Data has been added successfully.")


def main():
  topic = input("Enter the topic you want to search: ").strip()
  content = get_wikipedia_page(topic)
  if content:
    soup = content

    title = get_article_title(soup)
    summary = get_article_summary(soup)
    headings = get_article_headings(soup)[:5]
    links = get_article_links(soup)
    add_all_in_file(topic,title,summary,headings,links)
    add_article_in_json(topic,title,summary,headings,links)

    print(".....scrapping.....")

    print(f"Article Title: {title}")
    print(f"Article Summary: {summary}")
    print(f"Article Headings:")
    for heading in headings:
        print(f"- {heading}")
    print(f"Article Links: ")
    for link in links:
        print(f"- {link}")

main()
```
